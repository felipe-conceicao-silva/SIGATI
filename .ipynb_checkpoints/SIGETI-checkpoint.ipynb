# Organização do Sistema

O sistema foi dividido em módulos para deixar o código mais organizado e facilitar a manutenção. Cada arquivo possui uma função específica dentro da aplicação.

## Estrutura dos Arquivos

```text
inventario_seguranca/
├── main.py      ← menu principal e controle do sistema
├── ativos.py    ← cadastro e gerenciamento dos ativos
├── vulns.py     ← cadastro e gerenciamento das vulnerabilidades
├── dados.py     ← leitura e gravação dos arquivos JSON
├── ui.py        ← entrada de dados e exibição das informações
├── tipos.py     ← definições dos tipos, severidades e status
└── data/
    ├── ativos.json
    └── vulns.json
```

## O que cada arquivo faz

### main.py

É o arquivo principal do sistema.

Suas funções são:

- Inicializar os arquivos de dados;
- Carregar as informações salvas;
- Exibir o menu principal;
- Executar a ação escolhida pelo usuário.

### ativos.py

Contém as funções relacionadas aos ativos de TI:

- Cadastrar ativo;
- Consultar ativo;
- Atualizar ativo;
- Remover ativo;
- Listar ativos.

### vulns.py

Contém as funções relacionadas às vulnerabilidades:

- Cadastrar vulnerabilidade;
- Visualizar vulnerabilidades de um ativo;
- Atualizar status ou severidade.

### dados.py

Responsável por salvar e carregar os dados do sistema usando arquivos JSON.

Também cria automaticamente os arquivos necessários quando o programa é executado pela primeira vez.

### ui.py

Reúne as funções de interação com o usuário:

- Leitura de dados;
- Validação das entradas;
- Exibição de menus;
- Formatação das informações na tela.

### tipos.py

Define os valores permitidos pelo sistema utilizando `Enum`, como:

- Tipos de ativos;
- Níveis de severidade;
- Status das vulnerabilidades.

## Fluxo do Sistema

```text
main.py
│
├── Carrega os dados
├── Exibe o menu
├── Lê a opção escolhida
│
└── Chama a função correspondente
    ├── ativos.py
    └── vulns.py
```

## Resumo

A divisão em módulos deixa o código mais organizado e fácil de entender. Cada arquivo possui uma responsabilidade específica, evitando que toda a lógica fique concentrada em um único lugar.
