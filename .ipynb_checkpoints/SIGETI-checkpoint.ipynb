Este notebook apresenta a implementação completa do **SIGETI**, um sistema CRUD em Python para gerenciamento de ativos de TI e vulnerabilidades associadas.

## Criando o Repósitorio do Projeto

```bash
git init
git add .
git commit -m "estrutura inicial do projeto"
git checkout -b teste
```

Primeiro, inicializei um repositório Git na pasta do projeto. Em seguida, adicionei todos os arquivos e fiz o primeiro commit para salvar a estrutura inicial. Por fim, criei uma branch chamada teste.

Agora vou criar um repositorio

```bash
git remote add origin https://github.com/felipe-conceicao-silva/SIGATI.git
```

E enviar para o github

```bash
git push -u origin main
```
